```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import VotingClassifier 
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

# Logistic Regression
clf1 = LogisticRegression(
    random_state=1,
    solver='lbfgs'
)

# K-Nearest Neighbors
clf2 = KNeighborsClassifier(
    n_neighbors=5
)

# Decision Tree
clf3 = DecisionTreeClassifier(
    max_depth=3,
    random_state=1
)

print("Individual classifiers defined")
```

```python
# Import required libraries
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import VotingClassifier

# Load dataset
iris = load_iris()

X = iris.data
y = iris.target

# Split the data into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=1,
    stratify=y
)

# Scale the training and testing data
scaler = StandardScaler()

X_train_scaled_ensemble = scaler.fit_transform(X_train)
X_test_scaled_ensemble = scaler.transform(X_test)

# Create the Voting Classifier
voting_clf = VotingClassifier(
    estimators=[
        ('lr', clf1),
        ('knn', clf2),
        ('dt', clf3)
    ],
    voting='hard'
)

# Train the Voting Classifier
voting_clf.fit(
    X_train_scaled_ensemble,    y_train
)

print("Voting Classifier (hard voting) trained successfully.")
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Make predictions
y_pred_ensemble = voting_clf.predict(
    X_test_scaled_ensemble
)

# Calculate accuracy
ensemble_accuracy = accuracy_score(
    y_test,
    y_pred_ensemble
)

# Generate classification report
ensemble_report = classification_report(
    y_test,
    y_pred_ensemble
)

print(f"Ensemble Model Accuracy: {ensemble_accuracy:.2f}")
print("Ensemble Model Classification Report:")
print(ensemble_report)
```
